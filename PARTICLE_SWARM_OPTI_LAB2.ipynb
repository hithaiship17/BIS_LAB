```python

import random
import math

def distance(city1, city2):
    x1, y1 = city1
    x2, y2 = city2
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)


def route_distance(route, cities):
    total = 0
    for i in range(len(route)):
        city1 = cities[route[i]]
        city2 = cities[route[(i + 1) % len(route)]]
        total += distance(city1, city2)
    return total


def create_particle(n):
    route = list(range(n))
    random.shuffle(route)
    return route


def mutate(route):
    new_route = route.copy()
    i, j = random.sample(range(len(route)), 2)
    new_route[i], new_route[j] = new_route[j], new_route[i]
    return new_route


def tsp_pso(cities, particles=30, iterations=200):
    n = len(cities)

    swarm = [create_particle(n) for _ in range(particles)]
    pbest = [route.copy() for route in swarm]

    gbest = min(
        pbest,
        key=lambda route: route_distance(route, cities)
    )

    for iteration in range(iterations):
        for i in range(particles):
            current = swarm[i]
            new_route = current.copy()

            if random.random() < 0.5:
                new_route = mutate(new_route)

            if random.random() < 0.7:
                new_route = gbest.copy()
                new_route = mutate(new_route)

            swarm[i] = new_route

            if route_distance(new_route, cities) < route_distance(pbest[i], cities):
                pbest[i] = new_route.copy()

            if route_distance(pbest[i], cities) < route_distance(gbest, cities):
                gbest = pbest[i].copy()

        if (iteration + 1) % 20 == 0:
            print(
                "Iteration:", iteration + 1,
                "Best distance:",
                round(route_distance(gbest, cities), 2)
            )

    return gbest


# Main program
cities = [
    (0, 0),
    (2, 6),
    (5, 3),
    (8, 7),
    (6, 1),
    (3, 0)
]

best_route = tsp_pso(cities, particles=30, iterations=200)
best_distance = route_distance(best_route, cities)

print("\nBest route:")

for city in best_route:
    print("City", city, "->", end=" ")

print("City", best_route[0])
print("Total Distance:", round(best_distance, 2))
```
