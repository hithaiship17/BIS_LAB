```python

import numpy as np

# Step 1: Define the objective function
def fitness(x):
    return -(x - 5)**2 + 25


# Step 2: Initialize parameters
N = 10
MAX_ITERATIONS = 100
TOLERANCE = 0.0001
LOWER_BOUND = -10
UPPER_BOUND = 10
LEARNING_RATE = 0.5

# Step 3: Initialize population randomly
cells = np.random.uniform(
    LOWER_BOUND, UPPER_BOUND, (N, N)
)

# Evaluate initial fitness
fitness_values = fitness(cells)

# Store global best solution
best_index = np.unravel_index(
    np.argmax(fitness_values), fitness_values.shape
)

best_x = cells[best_index]
best_fitness = fitness_values[best_index]

print("Initial Best X:", best_x)
print("Initial Best Fitness:", best_fitness)


# Step 4: Start iterations
for iteration in range(MAX_ITERATIONS):

    # Store old cell states
    old_cells = cells.copy()

    # Find the best neighbor for every cell
    best_neighbor = old_cells.copy()
    best_neighbor_fitness = fitness(best_neighbor)

    # Step 5: Check 8 neighboring cells
    for i in [-1, 0, 1]:
        for j in [-1, 0, 1]:

            if i == 0 and j == 0:
                continue

            neighbor = np.roll(
                np.roll(old_cells, i, axis=0),
                j, axis=1
            )

            neighbor_fitness = fitness(neighbor)

            # Select the best neighboring solution
            mask = neighbor_fitness > best_neighbor_fitness

            best_neighbor[mask] = neighbor[mask]
            best_neighbor_fitness[mask] = neighbor_fitness[mask]

    # Step 6: Update all cells simultaneously
    cells = (
        old_cells
        + LEARNING_RATE * (best_neighbor - old_cells)
    )

    # Keep values within search range
    cells = np.clip(cells, LOWER_BOUND, UPPER_BOUND)

    # Evaluate new fitness
    fitness_values = fitness(cells)

    # Step 7: Update global best solution
    current_index = np.unravel_index(
        np.argmax(fitness_values), fitness_values.shape
    )

    current_best_fitness = fitness_values[current_index]

    if current_best_fitness > best_fitness:
        best_fitness = current_best_fitness
        best_x = cells[current_index]

    # Step 8: Check convergence condition
    change = np.max(np.abs(cells - old_cells))

    if change < TOLERANCE:
        print("Converged at iteration:", iteration + 1)
        break

# Step 9: Display final result
print("\nFinal Results")
print("Best X =", best_x)
print("Maximum Function Value =", best_fitness)
print("Iterations =", iteration + 1)
```
