```python

import random

# Distance matrix between 4 cities
D = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

# ACO parameters
n = len(D)
num_ants = 10
iterations = 100
alpha = 1.0       # Pheromone importance
beta = 2.0        # Distance heuristic importance
rho = 0.5         # Evaporation rate
Q = 100           # Pheromone deposit constant

# Step 1: Initialize pheromone trails
pheromone = [
    [0.0 if i == j else 1.0 for j in range(n)]
    for i in range(n)
]

best_tour = None
best_distance = float("inf")

# Step 2: Run ACO iterations
for iteration in range(iterations):
    ant_tours = []

    # Step 3: Each ant constructs a tour
    for ant in range(num_ants):
        start = random.randrange(n)
        tour = [start]
        unvisited = set(range(n))
        unvisited.remove(start)

        while unvisited:
            current = tour[-1]
            candidates = list(unvisited)

            # Calculate weights for possible next cities
            weights = [
                (pheromone[current][city] ** alpha)
                * ((1.0 / D[current][city]) ** beta)
                for city in candidates
            ]

            # Choose next city probabilistically
            next_city = random.choices(
                candidates, weights=weights, k=1
            )[0]

            tour.append(next_city)
            unvisited.remove(next_city)

        # Step 4: Calculate complete tour distance
        distance = sum(
            D[tour[i]][tour[(i + 1) % n]]
            for i in range(n)
        )

        ant_tours.append((tour, distance))

        # Keep track of the best tour
        if distance < best_distance:
            best_distance = distance
            best_tour = tour[:]

    # Step 5: Evaporate pheromones
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

    # Step 6: Deposit pheromones
    for tour, distance in ant_tours:
        deposit = Q / distance

        for i in range(n):
            a = tour[i]
            b = tour[(i + 1) % n]
            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

# Step 7: Display the best tour
city_names = ["A", "B", "C", "D"]
route = [city_names[i] for i in best_tour]
route.append(route[0])

print("Best tour:", " -> ".join(route))
print("Minimum distance:", best_distance)
```
