```python

import random

p = input("Enter initial population: ").split()
n = len(p)

pc = 0.8
pm = 0.01
gen = 10

def fit(s):
    x = int(s, 2)
    return x * x

for g in range(gen):
    f = [fit(s) for s in p]
    total = sum(f)

    if g == 0:
        print("\nString\tX\tFitness\tProbability\tExpected count")
        for i in range(n):
            prob = f[i] / total if total else 0
            print(p[i], "\t", int(p[i], 2), "\t", f[i],
                  "\t", round(prob * 100, 2), "%",
                  "\t", round(prob * n, 2))

    parents = random.choices(
        p, weights=f if total else None, k=n
    )
    new = []

    for i in range(0, n, 2):
        a = parents[i]
        b = parents[(i + 1) % n]

        if random.random() < pc:
            k = random.randint(1, len(a) - 1)
            a, b = a[:k] + b[k:], b[:k] + a[k:]

        for child in (a, b):
            child = list(child)

            for j in range(len(child)):
                if random.random() < pm:
                    child[j] = '1' if child[j] == '0' else '0'

            new.append(''.join(child))

    p = new[:n]

best = max(p, key=lambda s: fit(s))

print("\nBest string:", best)
print("X =", int(best, 2))
print("Maximum fitness =", fit(best))

```
